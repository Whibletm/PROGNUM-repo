```python
#3.5
import numpy as np

#N = int(input("N:"))

def conv(N):
    x = np.arange(1,(N+1))
    for i in x:
        sum_ = np.sum(1/(i**2))    
    print(f"the sum when N = {N} is {sum_}")


N_100 = conv(100)
N_500 = conv(500)
N_1000 = conv(1000)

print("since the sum as N goes towards infinity get increasingly miniscule the series can be said to converge at 0")

#prove that N converges by taking limit as N approaches infinity




```
