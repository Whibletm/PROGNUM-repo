```python
#3.22
import numpy as np
from matplotlib import pyplot as plt

x = list(range(-5,6,1))  #makes list
x_arr = np.arange(-5,6,1)  #makes array
print("list x:", x)

#y=cosh

cosh = np.cosh(x)  #applies function
print("cosh of the list:", cosh)

plt.plot(x, cosh, marker='o', color='green')  #creates a plot of x and cosh
plt.title("Catenary Plot", fontsize = 16) #title
plt.ylabel("cosh(x)", fontsize = 13)
plt.xlabel("x", fontsize = 13)
plt.grid()
plt.show()

print(f"The result of the arrange is {x_arr}, and is the same as the list 'x'")

#run it on terminal
```
