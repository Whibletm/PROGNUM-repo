```python
#4.3
import numpy as np

print("Rules; rock beats scissors, scissors beats paper and paper beats rock, good luck!")
choice_h1 = input("Choose either rock(R), paper(P) or scissors(S):")
choice_h = choice_h1.upper()


random = np.random.randint(1, 3+1)

if random ==1:
    choice_c = 'R'
elif random ==2:
    choice_c = 'S'
else:
    choice_c = 'P'



if choice_c == 'R':
    print("Computer chooses rock")
    if choice_h == 'S':
        print("Computer wins :(")
    elif choice_h == 'P':
        print("You win!")
    else:
        print("its a tie")

if choice_c == 'S':
    print("Computer chooses scissors")
    if choice_h == 'R':
        print("You win!")
    elif choice_h == 'P':
        print("Computer wins:(")
    else:
        print("It's a tie")

if choice_c == 'P':
    print("Computer chooses paper")
    if choice_h == 'R':
        print("Computer wins :(")
    elif choice_h == 'S':
        print("You win!")
    else:
        print("It's a tie")
```
