```python
import math
x = [1,2]  #makes first two terms of fib
while len(x)<=99:    #0nly runs until 100th term
        a = x[-1] + x[-2]    #adds the last number in sequence with the second to last number
        x.append(a)    #appends this value to the list


print(f"the 100th term of the fibonacci sequence is", x[-1])  #prints last value
```
