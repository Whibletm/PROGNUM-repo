```python
night1 = [1,3,2,4,0,3,7,5,2,9,2,4,3]
night2 = [0,0,2,4,2,4,3,2,1,1,1,2,0]

night1_s = set(night1)
night2_s = set(night2)

inter = set.intersection(night1_s, night2_s)

if 5 in inter:
    print("5 is part of the intersection")
else:
    print("5 is not a part of the intersection")
    
if 5 in night1_s:
    print ("5 is part of night1")
else:
    print("5 is not a part of night1")
```
