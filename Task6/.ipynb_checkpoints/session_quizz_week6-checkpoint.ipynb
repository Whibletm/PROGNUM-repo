```python
#6.3
import numpy as np
from astropy.io import fits
from matplotlib.pyplot import show, figure
import matplotlib.pyplot as plt
from imageio import imwrite, imread


m101=fits.open('m101.fits')
dat = np.array(m101[0].data) 



slice_ = dat[0:20, 0:20]   #slices the top corner of pixels

slice_f = slice_.flatten()


n=0
for i in slice_f:           #loop that gets the square parts of the eq
   n+=float(i)**2

rms = np.sqrt((1/(20*20))*n)   #eq that gets the RMS
print(f"RMS: {rms}")



x_std=np.std(slice_)
x_mean = np.mean(slice_)
print(f"standard deviation: {x_std}, mean: {x_mean}")



#why does it differ from the RMS?           RMS tests deviation from 0, std goes form mean
#which quantify image noise best?         std quantifies random noise well    
```

```python
#6.5
import numpy as np
from astropy.io import fits
from matplotlib.pyplot import show, figure
import matplotlib.pyplot as plt
from imageio import imwrite, imread

noise0=fits.open('noisemap0.fits')
noise1=fits.open('noisemap1.fits')
noise2=fits.open('noisemap2.fits')

dat0 = np.array(noise0[0].data) 
dat1 = np.array(noise1[0].data) 
dat2 = np.array(noise2[0].data) 


std_0=np.std(dat0)
std_1=np.std(dat1)
std_2=np.std(dat2)

stds=[std_0, std_1, std_2]

m_0=np.mean(dat0)
m_1=np.mean(dat1)
m_2=np.mean(dat2)

stack = np.stack([dat0, dat1, dat2])      #remember [], stack expects tuples, default 0 axis



mean_im = np.mean(stack, axis=0)     #axis=0 thru the stack (z), axis1=(cols) top to bottom, axis2=rows (left to right)    ---> DOUBLE CHECK 

mean_m= np.mean(mean_im)      #mean of averaged stack image
std_m=np.std(mean_im)       #std of averaged stack image

print(f"mean: {mean_m}, standard deviation: {std_m}")


#calculating standard deviation manually
N=3   #length of value set

sum_s=0    #creating a list to append to
for i in stds:     #looping for sum of squares
    sum_s += i**2
 
std_calc=(1/N)*np.sqrt(sum_s)      #taking last value of list (total sum)

print(f"calculated standard deviation:{std_calc}")      #go over notebook30 reducing noise with stack of images
```

```python
#6.14

class Fibonacci():
    """Class for calculating Fibonacci sequence"""
    def __init__(self, N, M):
        self.N = N
        self.M = M
        self.x=[0,1]
       
   
    def Nterm(self):
         if (self.N) == 0:   #accounts for if user wants N=0 or N==1
            self.x = []
         elif self.N == 1: 
            self.x = 0
         else:
            while len(self.x) < self.N:  # creates a fibo sequence that is N long
                y = self.x[-1] + self.x[-2]
                self.x.append(y)
         return(self.x[-1])
    
    def Mdivisible(self):
        x_m = []
        for i in self.x:
            if (i) % self.M == 0:    #checks if element is divisible by M
                x_m.append(i)  #adds to list if yes
       
            
        return(x_m)

fibon = Fibonacci(100,7)
fibon_Nterm = fibon.Nterm()
fibon_Mdivisible = fibon.Mdivisible()
print(f"Test (N=100, M=7):")
print(f"100-th term: {fibon_Nterm}")
print(f"Terms divisible by 7: {fibon_Mdivisible}")


```
